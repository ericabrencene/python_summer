## Списки

Немного теории

*   список - маленький массив данных одного типа;
*   это изменяемый тип данных (в отличие от строки)

поэтому код

```
a = [1, 2, 3]
b = a
b.append(4)
print(a)
```
выведет [1, 2, 3, 4]
(как сохранить исходный список - обсудим на следующем урока)

*   у каждого элемента списка есть номер - индекс
*   если номер элемента - n, то индекс - n-1

Методы и функции, которые могут вам понадобиться

```python

string_new = list(string)
# превращает строку string в список из букв и других символов string_new
# здесь нельзя указать разделитель - каждый символ энивей будет отдельным элементом

len(list_name)
len(string_name)
# выводит количество символов строки или количество элементов списка

list_name[index]
# это способ указать на номер списка в строке

list_name[index_first:index_second]
# срез от какого-то одного индекса к другому
# важный момент: index_second берем на 1 больше, чем нам нужно, потому что тут не включительно

#есть еще такой вариант
list_name[index_first:index_second:step]
# на месте step можно указать шаг и выводить например каждый второй элемент


sentence_new = sentence.split()
# делает из строки sentence список sentence_new, деля по пробелам
# здесь можно указать разделитель:
.split('.') #будет делить по точкам
.split('. ') #по комбинации "точка + пробел"

list_name.append(element)
# добавляет элемент element в конец списка list_name

list_name.insert(index, element)
# добавляет элемент element на позицию index списка list_name

list_name.pop() # может работать двумя способами

last_element = list_name.pop()
# мы определяем последний элемент и заодно удаляем его

list_name.pop(index)
# удаляем элемент с индексом index


```
